```python
opportunities = [
    {
        "title": "AI for Marine Biodiversity",
        "text": """
        Research funding for artificial intelligence,
        computer vision and marine ecosystem monitoring.
        """
    },

    {
        "title": "Medical Image Analysis Grant",
        "text": """
        Research opportunity using deep learning
        for medical image diagnosis.
        """
    },

    {
        "title": "Arabic NLP Fellowship",
        "text": """
        Fellowship focused on large language models
        and natural language processing.
        """
    },

    {
        "title": "Ancient History Scholarship",
        "text": """
        Scholarship for historical research
        and archaeological studies.
        """
    }
]
```

```python
researcher_vector = embedder.encode(
    researcher_text
)


results = []


for opportunity in opportunities:

    vector = embedder.encode(
        opportunity["text"]
    )

    score = matcher.compute_similarity(
        researcher_vector,
        vector
    )

    results.append(
        {
            "title": opportunity["title"],
            "score": score
        }
    )
```

```python
results = sorted(
    results,
    key=lambda x: x["score"],
    reverse=True
)


for result in results:
    print(
        result["title"],
        "→",
        round(result["score"],3)
    )
```
