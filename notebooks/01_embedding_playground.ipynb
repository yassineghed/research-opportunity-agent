```python
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np
```

```python
MODEL_NAME = "sentence-transformers/all-MiniLM-L6-v2"

model = SentenceTransformer(MODEL_NAME)

print("Model loaded successfully!")
```

```python
researcher_text = """
Research Domain:
Artificial Intelligence

Research Interests:
Marine Conservation
Computer Vision

Skills:
Python
Deep Learning
YOLO
"""

opportunity_text = """
Funding opportunity for research in Artificial Intelligence,
Computer Vision and Marine Biodiversity Monitoring.
"""

unrelated_text = """
Scholarship for Medieval Literature and Ancient Greek Philosophy.
"""
```

```python
researcher_embedding = model.encode(researcher_text)

opportunity_embedding = model.encode(opportunity_text)

unrelated_embedding = model.encode(unrelated_text)
```

```python
print(type(researcher_embedding))
print()

print("Embedding shape:", researcher_embedding.shape)
print()

print("First 10 values:")
print(researcher_embedding[:10])
```

```python
similarity_good = cosine_similarity(
    [researcher_embedding],
    [opportunity_embedding]
)[0][0]

similarity_bad = cosine_similarity(
    [researcher_embedding],
    [unrelated_embedding]
)[0][0]

print(f"Relevant opportunity : {similarity_good:.3f}")
print(f"Unrelated opportunity: {similarity_bad:.3f}")
```
